```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("baboon.jpg", 0)

k = np.array([[1,2,1],
              [2,4,2],
              [1,2,1]]) / 16

out = np.zeros(img.shape)

for i in range(1, img.shape[0]-1):
    for j in range(1, img.shape[1]-1):
        out[i,j] = np.sum(img[i-1:i+2, j-1:j+2] * k)

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original")

plt.subplot(1,2,2)
plt.imshow(out, cmap='gray')
plt.title("Smoothed")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("baboon.jpg", 0)

Gx = np.array([[-1,0,1],
               [-2,0,2],
               [-1,0,1]])

Gy = np.array([[-1,-2,-1],
               [0,0,0],
               [1,2,1]])

x = np.zeros(img.shape)
y = np.zeros(img.shape)

for i in range(1, img.shape[0]-1):
    for j in range(1, img.shape[1]-1):
        a = img[i-1:i+2, j-1:j+2]
        x[i,j] = np.sum(a * Gx)
        y[i,j] = np.sum(a * Gy)

G = np.sqrt(x**2 + y**2)

plt.subplot(2,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original")


plt.subplot(2,2,2)
plt.imshow(G, cmap='gray')
plt.title("Sobel Edge")

plt.show()
```
